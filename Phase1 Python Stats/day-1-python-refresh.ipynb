## 1. Data Types Recap

```python
# A list - ordered, changeable
prices = [10, 20, 30]

# A dictionary - a single record, like a row of data
record = {"name": "Ali", "age": 25}

# A tuple - fixed, unchangeable
coordinates = (37.77, -122.41)

# A set - removes duplicates automatically
unique_ids = {101, 102, 102, 103}

print(prices)
print(record)
print(coordinates)
print(unique_ids)
```

## 2. List Comprehensions - Worked Examples

```python
# Traditional loop way
squares = []
for x in range(10):
    squares.append(x**2)
print(squares)
```

```python
# Same result, list comprehension way
squares = [x**2 for x in range(10)]
print(squares)
```

```python
# List comprehension with a filter condition
evens = [x for x in range(20) if x % 2 == 0]
print(evens)
```

## 3. Lambda Functions - Worked Examples

```python
# A lambda function
square = lambda x: x**2
print(square(5))
```

```python
# Using a lambda with map()
numbers = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
```

```python
# Using a lambda with filter()
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
```
