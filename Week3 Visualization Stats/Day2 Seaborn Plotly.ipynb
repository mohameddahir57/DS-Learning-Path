```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

np.random.seed(42)

# Sample DataFrame used throughout this notebook
df = pd.DataFrame({
    "age": np.random.randint(18, 60, 150),
    "income": np.random.normal(50000, 15000, 150),
    "spending_score": np.random.randint(1, 100, 150),
    "city": np.random.choice(["Mogadishu", "Hargeisa", "Kismayo"], 150)
})
df.head()
```

---
## 1. Seaborn Basics - Worked Examples

```python
sns.histplot(data=df, x="age", bins=15)   # histogram, Seaborn style
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(data=df, x="city", y="age")   # box plot - median, spread, and outliers per category
plt.title("Age Spread by City")
plt.show()
```

---
## 2. Heatmaps - Worked Example

```python
correlation_matrix = df.corr(numeric_only=True)   # correlation between all numeric columns
print(correlation_matrix)

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")   # annot=True shows the numbers
plt.title("Correlation Heatmap")
plt.show()
```

---
## 3. Pairplots - Worked Example

```python
sns.pairplot(df, hue="city")   # scatter plot for every pair of numeric columns, colored by city
plt.show()
```

---
## 4. Plotly Interactive Charts - Worked Example

```python
fig = px.scatter(df, x="age", y="income", color="city", title="Age vs Income by City")
fig.show()
```

## 5. Exercises

```python
# 1. SOLVED - Box plot comparing spending_score across cities
sns.boxplot(data=df, x="city", y="spending_score", palette="Set2")
plt.title("Spending Score Spread by City")
plt.show()
```

```python
# 2. SOLVED - Build a DataFrame with 4 numeric columns, plot a correlation heatmap
sample = pd.DataFrame({
    "hours_studied": np.random.uniform(0, 10, 100),
    "attendance": np.random.uniform(50, 100, 100),
    "sleep_hours": np.random.uniform(4, 9, 100),
})
sample["exam_score"] = (sample["hours_studied"] * 5
                         + sample["attendance"] * 0.3
                         + np.random.normal(0, 5, 100))

corr = sample.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Between Study Habits and Exam Score")
plt.show()
```

```python
# 3. SOLVED - DataFrame with 3 numeric columns + 1 categorical column, pairplot with hue
pair_df = pd.DataFrame({
    "height_cm": np.random.normal(165, 10, 90),
    "weight_kg": np.random.normal(65, 12, 90),
    "age": np.random.randint(18, 50, 90),
    "group": np.random.choice(["A", "B", "C"], 90)
})

sns.pairplot(pair_df, hue="group")
plt.show()
```

```python
# 4. SOLVED - Interactive Plotly scatter, colored by category
fig = px.scatter(
    df, x="spending_score", y="income", color="city",
    title="Spending Score vs Income by City",
    hover_data=["age"]
)
fig.show()
```
