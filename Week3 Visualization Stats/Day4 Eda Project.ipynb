**Steps:**
1. Load & first look
2. Missing values & duplicates
3. Descriptive statistics
4. Distributions (histograms)
5. Relationships (heatmap/pairplot)
6. Categorical comparisons (bar/box plots)

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

# Simulated dataset: students' study habits and exam performance
n = 200
df = pd.DataFrame({
    "student_id": range(1, n + 1),
    "hours_studied": np.random.uniform(0, 10, n),
    "attendance_pct": np.random.uniform(50, 100, n),
    "sleep_hours": np.random.normal(7, 1.3, n),
    "exam_score": None,   # filled in below, based on the other columns
    "study_group": np.random.choice(["Morning", "Evening", "Weekend"], n)
})
df["exam_score"] = (df["hours_studied"] * 5 + df["attendance_pct"] * 0.3
                     + np.random.normal(0, 8, n)).clip(0, 100)

# Introduce a few missing values and a duplicate row, to make the cleaning step meaningful
df.loc[5:8, "sleep_hours"] = np.nan
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)   # duplicate the first row
```

---
## Step 1: Load & First Look

```python
print(df.shape)
df.head()
```

```python
df.info()
```

---
## Step 2: Missing Values & Duplicates

```python
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Fill missing sleep_hours with the column mean, drop the duplicate row
df["sleep_hours"] = df["sleep_hours"].fillna(df["sleep_hours"].mean())
df = df.drop_duplicates()

print("After cleaning -> missing values:", df.isnull().sum().sum(), "| duplicates:", df.duplicated().sum())
```

---
## Step 3: Descriptive Statistics

```python
df.describe()
```

```python
print("Exam score -> mean:", df["exam_score"].mean(), "| median:", df["exam_score"].median(),
      "| std:", df["exam_score"].std())
```

---
## Step 4: Distributions

```python
sns.histplot(df["exam_score"], bins=20, kde=True)
plt.title("Distribution of Exam Scores")
plt.show()

sns.histplot(df["hours_studied"], bins=20, kde=True)
plt.title("Distribution of Hours Studied")
plt.show()
```

---
## Step 5: Relationships

```python
numeric_cols = ["hours_studied", "attendance_pct", "sleep_hours", "exam_score"]
corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[numeric_cols + ["study_group"]], hue="study_group")
plt.show()
```

---
## Step 6: Categorical Comparisons

```python
group_counts = df["study_group"].value_counts()
plt.bar(group_counts.index, group_counts.values)
plt.title("Number of Students per Study Group")
plt.show()

sns.boxplot(data=df, x="study_group", y="exam_score")
plt.title("Exam Score Spread by Study Group")
plt.show()
```

---
## Step 7: Observations

- Exam scores are roughly normally distributed around the mid-60s to 70s, with a standard deviation of around 8-10 points - no extreme outliers after cleaning.
- The correlation heatmap shows **hours_studied** has the strongest positive correlation with **exam_score**, confirming the (simulated) relationship that more study time is associated with higher scores.
- **sleep_hours** shows a much weaker correlation with exam_score than hours_studied or attendance - in this dataset, sleep doesn't appear to matter as much as study time, which is worth noting as a finding rather than assuming beforehand.
- The box plot across study groups shows roughly similar exam score spread across Morning/Evening/Weekend groups, suggesting **study_group** alone isn't a strong predictor of performance - **hours_studied** matters more than *when* someone studies.
- Before cleaning, there were a few missing sleep_hours values and one duplicate row; both were handled in Step 2, which matters because leaving the duplicate in would have double-counted that student's influence on every summary statistic above.
