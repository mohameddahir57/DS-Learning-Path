```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
```

---
## 1. Mean, Median, Mode - Worked Example

```python
data = np.array([10, 12, 12, 15, 18, 95])   # 95 is a clear outlier

print("Mean:", np.mean(data))       # pulled upward by the outlier
print("Median:", np.median(data))   # barely affected by the outlier
print("Mode:", stats.mode(data, keepdims=True).mode[0])   # most frequent value
```

---
## 2. Variance & Standard Deviation - Worked Example

```python
print("Variance:", np.var(data))   # average of squared differences from the mean
print("Std Dev:", np.std(data))    # square root of variance - back in original units
```

---
## 3. Distributions - Worked Example

```python
normal_data = np.random.normal(50, 10, 1000)   # symmetric bell curve
skewed_data = np.random.exponential(10, 1000)   # right-skewed, long tail to the right

plt.hist(normal_data, bins=30, alpha=0.6, label="Normal")
plt.hist(skewed_data, bins=30, alpha=0.6, label="Right-skewed")
plt.legend()
plt.title("Normal vs Right-Skewed Distribution")
plt.show()
```

## 4. Exercises

```python
# 1. SOLVED - Manual mean/std calculation vs built-in functions
values = np.array([4, 8, 6, 5, 3, 7])

# Manual mean: sum of values / count of values
manual_mean = sum(values) / len(values)

# Manual std: sqrt of (average of squared differences from the mean)
manual_variance = sum((x - manual_mean) ** 2 for x in values) / len(values)
manual_std = manual_variance ** 0.5

print("Manual mean:", manual_mean, "  | np.mean:", np.mean(values))
print("Manual std:", manual_std, "  | np.std:", np.std(values))
# Both match - confirms the manual formula is correct
```

```python
# 2. SOLVED - Mean vs median, before and after adding an outlier
scores = np.array([70, 72, 68, 75, 71])
print("Before outlier -> mean:", np.mean(scores), "| median:", np.median(scores))

scores_with_outlier = np.append(scores, 10)   # one very low outlier score
print("After outlier  -> mean:", np.mean(scores_with_outlier), "| median:", np.median(scores_with_outlier))
# The mean dropped noticeably; the median barely moved - this shows the mean's sensitivity to outliers
```

```python
# 3. SOLVED - Find the mode of a dataset with a clearly repeated value
shoe_sizes = np.array([40, 41, 42, 42, 42, 43, 44])
mode_result = stats.mode(shoe_sizes, keepdims=True)
print("Mode:", mode_result.mode[0], "| appeared", mode_result.count[0], "times")
```

```python
# 4. SOLVED - Two datasets, same mean, different standard deviation
group_consistent = np.array([68, 70, 69, 71, 70, 72])    # tightly clustered
group_variable = np.array([40, 95, 55, 85, 60, 85])       # same-ish mean, much more spread out

print("Consistent group -> mean:", np.mean(group_consistent), "| std:", np.std(group_consistent))
print("Variable group   -> mean:", np.mean(group_variable), "| std:", np.std(group_variable))
# Practical meaning: both groups average around the same score, but the 'variable' group is far
# less consistent/predictable - some individuals score very high, others very low
```

```python
# 5. SOLVED - Detect skew using mean vs median
right_skewed = np.random.exponential(10, 1000)

print("Mean:", np.mean(right_skewed))
print("Median:", np.median(right_skewed))
# The mean is noticeably HIGHER than the median - this confirms the data is right-skewed:
# a handful of large values in the long tail are pulling the mean upward,
# while the median (less affected by extreme values) stays lower, closer to the 'typical' value
```
