```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)   # reproducible random data throughout this notebook
```

---
## 1. Line Chart - Worked Example

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 170, 140, 200]

plt.plot(months, sales)         # draws a line connecting each point in order
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

---
## 2. Bar Chart - Worked Example

```python
cities = ["Mogadishu", "Hargeisa", "Kismayo"]
population = [2500, 800, 400]

plt.bar(cities, population)   # one rectangular bar per category
plt.title("Population by City (thousands)")
plt.show()
```

---
## 3. Histogram - Worked Example

```python
scores = np.random.normal(70, 10, 200)   # 200 simulated exam scores, mean=70, std=10

plt.hist(scores, bins=15)   # groups values into 15 ranges (bins) and counts how many fall in each
plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

---
## 4. Scatter Plot - Worked Example

```python
hours_studied = np.random.uniform(0, 10, 50)
exam_score = hours_studied * 6 + np.random.normal(0, 8, 50)   # score roughly depends on hours studied

plt.scatter(hours_studied, exam_score)   # one dot per (x, y) data point
plt.title("Hours Studied vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

## 5. Exercises

```python
# 1. SOLVED - Line chart of 6 months of made-up revenue
revenue_months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [5000, 5400, 4800, 6100, 7000, 6800]

plt.plot(revenue_months, revenue, marker="o")   # marker='o' adds a dot at each data point
plt.title("Revenue - First Half of the Year")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.show()
```

```python
# 2. SOLVED - Bar chart comparing test scores across 5 subjects
subjects = ["Math", "Science", "English", "History", "Art"]
avg_scores = [78, 85, 72, 68, 91]

plt.bar(subjects, avg_scores, color="steelblue")
plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.show()
```

```python
# 3. SOLVED - Histogram of 300 random normal values, 20 bins
data = np.random.normal(50, 15, 300)

plt.hist(data, bins=20, color="seagreen")
plt.title("Distribution of 300 Random Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4. SOLVED - Scatter plot of two related random variables
advertising_spend = np.random.uniform(100, 1000, 60)
sales_result = advertising_spend * 0.8 + np.random.normal(0, 100, 60)   # sales roughly follows spend

plt.scatter(advertising_spend, sales_result, color="darkorange")
plt.title("Advertising Spend vs Sales")
plt.xlabel("Advertising Spend ($)")
plt.ylabel("Sales ($)")
plt.show()
```
