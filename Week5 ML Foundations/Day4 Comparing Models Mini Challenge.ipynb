```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import train_test_split

np.random.seed(42)
```

---
## Week 5 Mini Challenge (Solved)

We build ONE dataset (hours studied -> exam score), then use it in two different ways:
- As raw scores, for Linear Regression (predicting a number)
- As pass/fail, for Logistic Regression (predicting a category)

```python
# Step 1: Create the dataset
hours_studied = np.random.uniform(0, 10, 150)
exam_score = (hours_studied * 7 + np.random.normal(0, 10, 150)).clip(0, 100)

df = pd.DataFrame({"hours_studied": hours_studied, "exam_score": exam_score})
df["passed"] = (df["exam_score"] >= 60).astype(int)   # pass/fail version
df.head()
```

```python
# Step 2: Linear Regression - predicting the EXACT score
X_reg = df[["hours_studied"]]
y_reg = df["exam_score"]

Xr_train, Xr_test, yr_train, yr_test = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

linear_model = LinearRegression()
linear_model.fit(Xr_train, yr_train)

linear_score = linear_model.score(Xr_test, yr_test)
print("Linear Regression score (R-squared):", linear_score)
```

```python
# Step 3: Logistic Regression - predicting PASS/FAIL
X_clf = df[["hours_studied"]]
y_clf = df["passed"]

Xc_train, Xc_test, yc_train, yc_test = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42)

logistic_model = LogisticRegression()
logistic_model.fit(Xc_train, yc_train)

logistic_score = logistic_model.score(Xc_test, yc_test)
print("Logistic Regression score (accuracy):", logistic_score)
```

**Step 4: SOLVED - Why can't we directly compare these two scores?**

The Linear Regression score (R-squared) measures how well the model predicts the EXACT numeric score.
The Logistic Regression score (accuracy) measures how often the model correctly guesses the CATEGORY (pass or fail).
These are two completely different measurements, answering two different questions - a high R-squared and a high accuracy both mean 'good', but they are not on the same scale or measuring the same thing. We can only fairly compare Linear Regression to other Linear Regression models, and Logistic Regression to other Logistic Regression models.

```python
# Step 5: Use both models to predict for a NEW student who studied 6 hours
new_student = [[6]]

predicted_score = linear_model.predict(new_student)
predicted_pass_fail = logistic_model.predict(new_student)
predicted_probability = logistic_model.predict_proba(new_student)

print("Linear Regression predicts exam score:", predicted_score[0])
print("Logistic Regression predicts pass/fail (1=pass):", predicted_pass_fail[0])
print("Logistic Regression probability (fail, pass):", predicted_probability[0])
```
