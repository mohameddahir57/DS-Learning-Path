---
## Example 'dirty' DataFrame (for the worked examples below)

```python
df_example = pd.DataFrame({
    "name": ["Ali", "Halimo", "Amina", "Ali", np.nan],
    "age": [25, np.nan, 28, 25, 40],
    "city": ["Mogadishu", "Hargeisa", np.nan, "Mogadishu", "Kismayo"]
})
print(df_example)
```

---
## 1. Finding Missing Values - Worked Examples

```python
print(df_example.isnull())               # True/False for every cell - True means missing
print(df_example.isnull().sum())         # count of missing values PER COLUMN
print(df_example.isnull().sum().sum())   # total missing values in the whole DataFrame
```

---
## 2. Handling Missing Values - Worked Examples

```python
dropped = df_example.dropna()                    # drop ANY row with at least one missing value
print(dropped)

dropped_subset = df_example.dropna(subset=["age"])   # drop rows only where 'age' is missing
print(dropped_subset)
```

```python
filled_age = df_example.copy()
filled_age["age"] = filled_age["age"].fillna(filled_age["age"].mean())  # fill missing ages with the mean
print(filled_age)

filled_city = df_example.copy()
filled_city["city"] = filled_city["city"].fillna("Unknown")  # fill missing text with a placeholder
print(filled_city)
```

---
## 3. Handling Duplicates - Worked Examples

```python
print(df_example.duplicated())          # True/False per row - True means it's a duplicate
print(df_example.duplicated().sum())    # count of duplicate rows

no_dupes = df_example.drop_duplicates()  # remove duplicates, keep the FIRST occurrence
print(no_dupes)
```

---
## 4. Data Type Conversion - Worked Examples

```python
df_types = pd.DataFrame({
    "age_as_text": ["25", "30", "28"],
    "date_as_text": ["2024-01-01", "2024-02-15", "2024-03-20"]
})
print(df_types.info())   # notice both columns show up as 'object' (text)

df_types["age_as_text"] = df_types["age_as_text"].astype(int)         # convert to integer
df_types["date_as_text"] = pd.to_datetime(df_types["date_as_text"])   # convert to a real date type
print(df_types.info())   # now the types are correct
```

```python
# 1. Run .isnull().sum() to find missing values in each column

# isnull() marks missing cells as True.
# sum() counts the True values in each column.
missing_counts = df.isnull().sum()

# Print the number of missing values for every column.
print(missing_counts)

# Show only the columns that actually contain missing values.
print("Columns with missing values:")
print(missing_counts[missing_counts > 0])
```

```python
# 2. Print the total number of missing values in the whole dataset

# The first sum counts missing values in each column.
# The second sum adds those column totals together.
total_missing_values = df.isnull().sum().sum()

print("Total missing values:", total_missing_values)
```

```python
# 3. Create a cleaned copy with dropna() and count removed rows

# copy() protects the original DataFrame from being changed.
cleaned_df = df.dropna().copy()

# len() tells us how many rows are in a DataFrame.
original_row_count = len(df)
cleaned_row_count = len(cleaned_df)
removed_row_count = original_row_count - cleaned_row_count

print("Original rows:", original_row_count)
print("Cleaned rows:", cleaned_row_count)
print("Rows removed:", removed_row_count)
```

```python
# 4. Drop rows where one important column is missing

# Select the first column that has at least one missing value.
important_column = missing_counts[missing_counts > 0].index[0]

# subset tells dropna() to check only this one column.
cleaned_important_column = df.dropna(subset=[important_column])

print("Column checked:", important_column)
print(cleaned_important_column)
```

```python
# 5. Fill missing values in a numeric column with its mean

# Find numeric columns that have missing values.
numeric_missing_columns = df.select_dtypes(include="number").columns[
    df.select_dtypes(include="number").isnull().any()
]
numeric_column = numeric_missing_columns[0]

# Calculate the average of the available values in that column.
numeric_mean = df[numeric_column].mean()

# Make a copy, then fill missing values with the calculated mean.
filled_numeric_df = df.copy()
filled_numeric_df[numeric_column] = filled_numeric_df[numeric_column].fillna(numeric_mean)

print("Numeric column:", numeric_column)
print("Mean used to fill missing values:", numeric_mean)
print(filled_numeric_df[numeric_column])
```

```python
# 6. Fill missing values in a text column with "Unknown"

# Find text columns that contain at least one missing value.
text_data = df.select_dtypes(exclude="number")
text_missing_columns = text_data.columns[text_data.isnull().any()]
text_column = text_missing_columns[0]

# Make a copy so the original DataFrame stays unchanged.
filled_text_df = df.copy()
filled_text_df[text_column] = filled_text_df[text_column].fillna("Unknown")

print("Text column:", text_column)
print(filled_text_df[text_column])
```

```python
# 7. Count duplicate rows

# duplicated() returns True for rows that repeat an earlier row.
# sum() counts how many True values there are.
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# 8. Remove duplicate rows and confirm the new row count

# drop_duplicates() keeps the first copy and removes repeated rows.
no_duplicate_df = df.drop_duplicates()

print("Rows before removing duplicates:", len(df))
print("Rows after removing duplicates:", len(no_duplicate_df))
print(no_duplicate_df)
```

```python
# 9. Find a text column containing numbers and convert it to a numeric type

# Look at text columns and test whether their values can become numbers.
text_columns = df.select_dtypes(include="object").columns
numeric_text_column = None

for column in text_columns:
    values_as_numbers = pd.to_numeric(df[column], errors="coerce")
    non_missing_values = df[column].notna()

    # A column is a good match when all its non-missing values are numeric.
    if non_missing_values.any() and values_as_numbers[non_missing_values].notna().all():
        numeric_text_column = column
        break

if numeric_text_column is not None:
    converted_df = df.copy()

    # Convert the text numbers into real numbers.
    converted_df[numeric_text_column] = pd.to_numeric(
        converted_df[numeric_text_column], errors="coerce"
    )

    print("Column converted:", numeric_text_column)
    print("Data types after conversion:")
    print(converted_df[[numeric_text_column]].dtypes)
else:
    # Some datasets do not have a numeric column stored as text.
    print("No text column containing only numbers was found to convert.")
    print("Use df.info() to inspect the available data types.")
```
