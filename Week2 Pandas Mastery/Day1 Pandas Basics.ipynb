## 1. Series vs DataFrame - Worked Examples

```python
# Series -> a single column of data, with a label (index) for each value
ages = pd.Series([25, 30, 35], name="age")
print(ages)
```

```python
# DataFrame -> a full table made of multiple Series (columns) sharing the same index
df_example = pd.DataFrame({
    "name": ["Ali", "Halimo", "Amina"],
    "age": [25, 30, 28],
    "city": ["Mogadishu", "Hargeisa", "Kismayo"]
})
print(df_example)
```

## 4. Exercise (8 questions)

Pick a beginner-friendly dataset from Kaggle (Titanic or Iris are great starting points), upload it, and complete the questions below.

```python
# 1. Load your dataset into a DataFrame using pd.read_csv()
# First, upload your CSV file to Colab.
# Then use the exact file name here.
# Example: if your file is called "titanic.csv", use that name.
df = pd.read_csv("titanic.csv")

# This line reads the CSV file and saves it in a table called df.
# df is like an Excel sheet inside Python.
```

```python
# 2. Show the first 5 rows, then show the first 10 rows
# .head() shows the first 5 rows of the table
print(df.head())

# .head(10) shows the first 10 rows
print(df.head(10))


# This helps us see what the data looks like at the beginning.
```

```python
# 3. Show the last 5 rows of the dataset
# .tail() shows the last rows of the table
print(df.tail())


# This helps us see the end of the dataset.
```

```python
# 4. Use .info() and count how many columns have missing values
# .info() tells us the columns, data types, and missing data
# We can also count which columns have at least one empty value
df.info()

missing_columns = df.columns[df.isnull().any()].tolist()
print("Columns with missing values:", missing_columns)
print("Number of columns with missing values:", len(missing_columns))

# The number printed above tells us how many columns have missing values.
# Example: if it prints 3, then 3 columns are missing some values.
```

```python
# 5. Use .describe() and find which numeric column has the largest standard deviation
# .describe() gives a quick summary of the numbers in the dataset
print(df.describe())

# Pick only the numeric columns
numeric_columns = df.select_dtypes(include=['number'])

# Find the standard deviation of each numeric column
std_values = numeric_columns.std()
print(std_values)

# Show the column with the biggest standard deviation
largest_std_column = std_values.idxmax()

print("Numeric column with the largest std:", largest_std_column)

# A larger standard deviation means bigger spread in the values.
# This tells us which number column varies the most.
```

```python
# 6. Print df.shape to see how many rows and columns the dataset has
# df.shape shows the size of the table
print(df.shape)

rows = df.shape[0]
columns = df.shape[1]
print("Rows:", rows)
print("Columns:", columns)

# Example: if it prints (891, 12), it means 891 rows and 12 columns.
# This dataset has the number of rows and columns shown above.
```

```python
# 7. Show the column names and separate numeric and categorical columns
# df.columns shows all the names of the columns
print(df.columns)

# Numeric columns are columns with numbers
numeric_cols = df.select_dtypes(include=['number']).columns.tolist()

# Categorical columns are columns with text or labels
categorical_cols = df.select_dtypes(exclude=['number']).columns.tolist()

print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)


# Numeric columns usually contain values like age, price, score, or marks.
# Categorical columns usually contain words like sex, class, city, or country.
```

```python
# 8. Create your own Series and DataFrame manually
# A Series is one column of numbers or words
numbers_series = pd.Series([10, 20, 30, 40, 50], name="numbers")
print(numbers_series)

# A DataFrame is a table with rows and columns
student_df = pd.DataFrame({
    "student_name": ["Ali", "Hodan", "Amina"],
    "score": [88, 92, 85],
    "city": ["Mogadishu", "Hargeisa", "Kismayo"]
})
print(student_df)

# This example shows how a small DataFrame is made in Python.

# One column can be text, and another column can be numbers.
```
