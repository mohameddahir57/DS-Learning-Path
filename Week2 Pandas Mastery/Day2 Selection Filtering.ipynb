## 1. Selecting Columns - Worked Examples
(Using the small example DataFrame below so you can see the output right away)

```python
df_example = pd.DataFrame({
    "name": ["Ali", "Halimo", "Amina", "Yusuf"],
    "age": [25, 30, 28, 35],
    "city": ["Mogadishu", "Hargeisa", "Kismayo", "Mogadishu"]
})

print(df_example["age"])              # select ONE column -> returns a Series
print(df_example[["name", "age"]])    # select MULTIPLE columns -> returns a DataFrame (double brackets)
```

---
## 2. .loc and .iloc - Worked Examples

```python
print(df_example.loc[0])              # the row labeled 0
print(df_example.loc[0, "name"])      # value at row label 0, column 'name'
print(df_example.loc[0:2, "name"])    # rows 0 through 2 INCLUSIVE, column 'name'
```

```python
print(df_example.iloc[0])              # first row, by position
print(df_example.iloc[0, 1])            # first row, second column, by position
print(df_example.iloc[0:3, 0:2])        # first 3 rows, first 2 columns (EXCLUSIVE of end position)
```

---
## 3. Conditional Filtering - Worked Examples

```python
adults = df_example[df_example["age"] >= 18]                  # rows where age >= 18
print(adults)

mogadishu_only = df_example[df_example["city"] == "Mogadishu"]  # rows where city equals 'Mogadishu'
print(mogadishu_only)
```

```python
# Combined condition: use & (and) / | (or), each condition wrapped in its own parentheses
young_adults = df_example[(df_example["age"] >= 18) & (df_example["age"] <= 30)]
print(young_adults)
```

---
## 4. Sorting - Worked Examples

```python
print(df_example.sort_values("age"))                    # ascending by age
print(df_example.sort_values("age", ascending=False))   # descending by age
print(df_example.sort_values(["city", "age"]))          # sort by city first, then age
```

```python
# 6. Filter the dataset using one condition

# Calculate the average of the first numeric column.
average_value = df[numeric_column].mean()

# Keep only rows where the value is greater than the average.
rows_above_average = df[df[numeric_column] > average_value]

# Print the rows that match the condition.
print(rows_above_average)
```

```python
# 7. Filter the dataset using two conditions with &

# Find the smallest and largest values in the numeric column.
smallest_value = df[numeric_column].min()
largest_value = df[numeric_column].max()

# Keep values greater than the smallest value and less than the largest value.
# Each condition needs parentheses when using &.
rows_between_min_and_max = df[
    (df[numeric_column] > smallest_value) &
    (df[numeric_column] < largest_value)
]

# Print the rows that match both conditions.
print(rows_between_min_and_max)
```
