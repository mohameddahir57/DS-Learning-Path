## 1. GroupBy - Worked Examples

```python
df_example = pd.DataFrame({
    "city": ["Mogadishu", "Hargeisa", "Mogadishu", "Kismayo", "Hargeisa"],
    "gender": ["F", "M", "M", "F", "F"],
    "age": [25, 30, 28, 35, 22]
})

print(df_example.groupby("city")["age"].mean())   # average age, grouped by city
print(df_example.groupby("city")["age"].sum())    # total age, grouped by city
print(df_example.groupby("city").size())           # count of rows per city
```

```python
# Multiple aggregations at once
print(df_example.groupby("city")["age"].agg(["mean", "sum", "count"]))

# Group by multiple columns
print(df_example.groupby(["city", "gender"])["age"].mean())
```

## 2. Merging DataFrames - Worked Examples

```python
orders = pd.DataFrame({"order_id": [1, 2, 3], "customer_id": [101, 102, 101]})
customers = pd.DataFrame({"customer_id": [101, 102], "name": ["Ali", "Halimo"]})

merged = orders.merge(customers, on="customer_id")   # combine on the shared key column
print(merged)
```

## 3. Concatenating DataFrames - Worked Example

```python
jan_data = pd.DataFrame({"sales": [100, 200]})
feb_data = pd.DataFrame({"sales": [150, 250]})

combined_rows = pd.concat([jan_data, feb_data])            # stack ROWS on top of each other
print(combined_rows)

combined_cols = pd.concat([jan_data, feb_data], axis=1)    # stack side by side as new COLUMNS
print(combined_cols)
```

```python
# 2. Calculate mean, sum, and count for the same groups

# agg() lets us calculate several summaries at the same time.
summary_by_category = df.groupby(categorical_column)[numeric_column].agg(
    ["mean", "sum", "count"]
)

# mean = average, sum = total, and count = number of rows.
print(summary_by_category)
```

```python
# 3. Group by two categorical columns and calculate a numeric average

# Choose a second categorical column that is different from the first one.
second_categorical_column = categorical_columns[1]

# The two columns together create smaller, more specific groups.
mean_by_two_categories = df.groupby(
    [categorical_column, second_categorical_column]
)[numeric_column].mean()

print("Grouped by:", categorical_column, "and", second_categorical_column)
print(mean_by_two_categories)
```

## 6. Exercise - Part B: Merging & Concatenating (3 questions)

```python
# 4. Merge the sample orders and customers tables

# customer_id is the shared key found in both tables.
# An inner merge keeps rows that have a matching customer_id.
merged_orders = orders.merge(customers, on="customer_id", how="inner")

print(merged_orders)
```

```python
# 5. Add an order for a customer who is not in the customers table

# copy() keeps the original orders table unchanged.
orders_with_new_order = orders.copy()

# Customer 999 does not appear in the customers table.
extra_order = pd.DataFrame({"order_id": [4], "customer_id": [999]})
orders_with_new_order = pd.concat(
    [orders_with_new_order, extra_order],
    ignore_index=True
)

# A left merge keeps every row from the orders table.
left_merged_orders = orders_with_new_order.merge(
    customers,
    on="customer_id",
    how="left"
)

# The new order remains, but its customer name is missing because no match exists.
print(left_merged_orders)
```

```python
# 6. Concatenate two DataFrames with the same columns

# Create a small table for each month.
january_sales = pd.DataFrame({
    "month": ["January", "January"],
    "sales": [100, 200]
})
february_sales = pd.DataFrame({
    "month": ["February", "February"],
    "sales": [150, 250]
})

# concat() with the default axis=0 stacks the rows vertically.
combined_sales = pd.concat(
    [january_sales, february_sales],
    ignore_index=True
)

print(combined_sales)
```

## 7. Exercise - Part C: Week 2 Mini Challenge (Sales/Data Analysis Pipeline)

Combines all four days of Week 2. Follow the steps below **in order**.

```python
# Step 1: Use the cleaned dataset already loaded in df

# copy() creates a separate DataFrame for the mini challenge.
pipeline_df = df.copy()

print("Rows available for the mini challenge:", len(pipeline_df))
```

```python
# Step 2: Filter the data using a meaningful numeric condition

# Calculate the average of the numeric column.
pipeline_average = pipeline_df[numeric_column].mean()

# Keep rows where the numeric value is greater than or equal to the average.
filtered_pipeline_df = pipeline_df[
    pipeline_df[numeric_column] >= pipeline_average
].copy()

print("Rows after filtering:", len(filtered_pipeline_df))
print(filtered_pipeline_df)
```

```python
# Step 3: Group the filtered data and calculate two summaries

# Group the filtered rows by the same categorical column.
# mean shows the average value, and count shows how many rows are in each group.
grouped_pipeline = filtered_pipeline_df.groupby(categorical_column)[
    numeric_column
].agg(["mean", "count"])

print(grouped_pipeline)
```

```python
# Step 4: Sort the grouped result from highest to lowest

# Sort by the mean column in descending order.
# The group with the highest average appears first.
sorted_pipeline = grouped_pipeline.sort_values(
    by="mean",
    ascending=False
)

print(sorted_pipeline)
```

**Step 5 summary:**

The table shows the average numeric value for each category after filtering the data. The first row has the highest average because the groups were sorted from highest to lowest. The `count` column also shows how many records were included in each category.
