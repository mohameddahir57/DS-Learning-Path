```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
```

---
## 1. A/B Testing - Example

```python
# Click rates: 1 = clicked, 0 = did not click
group_a_clicks = [1, 0, 0, 1, 0, 0, 1, 0, 0, 0]   # old button
group_b_clicks = [1, 1, 0, 1, 1, 0, 1, 1, 0, 1]   # new button

rate_a = sum(group_a_clicks) / len(group_a_clicks)
rate_b = sum(group_b_clicks) / len(group_b_clicks)

print("Group A click rate:", rate_a)
print("Group B click rate:", rate_b)

t_stat, p_value = stats.ttest_ind(group_a_clicks, group_b_clicks)
print("p-value:", p_value)

if p_value < 0.05:
    print("The difference is likely REAL - the new button performs differently")
else:
    print("Not enough proof yet - could be random chance")
```

## 2. Mini Project 1: Full EDA + A/B Test (Solved Example)

Simulated dataset: an online store testing two website designs (A and B), tracking how much each customer spent.

```python
# Step 1: Load the data (simulated here)
n = 100
df = pd.DataFrame({
    "customer_id": range(1, n + 1),
    "website_version": np.random.choice(["A", "B"], n),
    "amount_spent": None
})
# Make version B slightly better on average, so there's a real difference to find
df["amount_spent"] = df["website_version"].apply(
    lambda v: np.random.normal(50, 10) if v == "A" else np.random.normal(58, 10)
)
df.head()
```

```python
# Step 2: Short EDA
print(df.isnull().sum())      # check missing values
print(df["amount_spent"].describe())

df["amount_spent"].hist(bins=20)
plt.title("Amount Spent - All Customers")
plt.show()
```

```python
# Step 3: Split into the two groups we want to compare
group_a = df[df["website_version"] == "A"]["amount_spent"]
group_b = df[df["website_version"] == "B"]["amount_spent"]

print("Group A average spend:", group_a.mean())
print("Group B average spend:", group_b.mean())
```

**Step 4: Write the hypotheses**

- Null hypothesis (H0): website version A and website version B lead to the SAME average amount spent (no real difference)
- Alternative hypothesis (H1): website version A and website version B lead to DIFFERENT average amounts spent

```python
# Step 5: Run the t-test
t_stat, p_value = stats.ttest_ind(group_a, group_b)
print("p-value:", p_value)
```

**Step 6: Conclusion**

If the p-value above is less than 0.05, the difference in average spending between website A and website B is likely real, not random chance. In that case, the recommendation would be: switch to whichever version had the higher average spend, since it appears to genuinely perform better.

If the p-value is 0.05 or higher, we don't have enough proof yet that one version is actually better - the recommendation would be to collect more data (more customers) before making a decision, rather than switching based on a difference that might just be random chance.
