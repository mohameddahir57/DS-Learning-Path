```python
from scipy import stats
import numpy as np

np.random.seed(42)
```

---
## 1 & 2. Hypotheses and P-Value - Example

**Example question:** Does sleep affect test scores?

- Null hypothesis (H0): sleep does NOT affect test scores (any difference is random chance)
- Alternative hypothesis (H1): sleep DOES affect test scores

If the p-value from our test is small (below 0.05), we believe H1. If it's large, we stick with H0.

---
## 3. T-Test - Example

```python
# Test scores from two different classes
class_a = [70, 72, 68, 75, 71, 69]
class_b = [80, 85, 78, 82, 88, 79]

t_stat, p_value = stats.ttest_ind(class_a, class_b)   # compares the two groups

print("p-value:", p_value)

if p_value < 0.05:
    print("The difference is likely REAL (not random chance)")
else:
    print("The difference could be random chance")
```
