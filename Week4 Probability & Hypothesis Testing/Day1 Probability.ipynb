```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)   # same random numbers every time we run this
```

---
## 1. What Is Probability - Example

```python
# A coin has 2 sides. 1 way to get heads, out of 2 total.
prob_heads = 1 / 2
print("Probability of heads:", prob_heads)   # 0.5 means 50%
```

---
## 2. Simple Rules - Example

```python
# AND rule: multiply the two probabilities together
prob_two_heads = (1/2) * (1/2)
print("Probability of 2 heads in a row:", prob_two_heads)
```

```python
# OR rule: add the two probabilities together
prob_1_or_2 = (1/6) + (1/6)
print("Probability of rolling a 1 OR a 2:", prob_1_or_2)
```

---
## 3. Normal Distribution - Example

```python
# 1000 fake heights. Average = 165. Most values stay close to 165.
heights = np.random.normal(165, 8, 1000)

plt.hist(heights, bins=30)
plt.title("Heights - Normal Distribution")
plt.show()
```
