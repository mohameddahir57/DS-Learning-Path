```python
import numpy as np
from scipy import stats

np.random.seed(42)
```

---
## 1 & 2. Confidence Interval - Example

```python
scores = [70, 72, 68, 75, 71, 69, 74, 73]

mean = np.mean(scores)
sem = stats.sem(scores)   # standard error - how much the average might vary

confidence_interval = stats.t.interval(0.95, len(scores) - 1, loc=mean, scale=sem)

print("Average:", mean)
print("95% confidence interval:", confidence_interval)
# Simple meaning: we believe the TRUE average score is probably somewhere in this range
```

---
## 3. Type I and Type II Errors

- Type I Error = false alarm (we say there's a difference, but there really isn't)
- Type II Error = missed finding (we say there's no difference, but there really is one)

## 4. Practice (Solved) - 4 Questions

```python
# 1. SOLVED - 95% confidence interval for a small group of made-up prices
prices = [12, 15, 14, 18, 13, 16, 17]

price_mean = np.mean(prices)
price_sem = stats.sem(prices)
price_ci = stats.t.interval(0.95, len(prices) - 1, loc=price_mean, scale=price_sem)

print("Average price:", price_mean)
print("95% confidence interval:", price_ci)
```

**2. SOLVED - What does this result mean?**

We are 95% confident that the TRUE average price (if we had every possible price, not just this small sample) falls somewhere inside the range printed above. The range is fairly wide because we only used 7 prices - a small sample size.

```python
# 3. SOLVED - Same idea, but with a MUCH bigger group (50 numbers)
big_group = np.random.normal(15, 2, 50)   # 50 fake prices, average around 15

big_mean = np.mean(big_group)
big_sem = stats.sem(big_group)
big_ci = stats.t.interval(0.95, len(big_group) - 1, loc=big_mean, scale=big_sem)

print("Average:", big_mean)
print("95% confidence interval:", big_ci)
# This range should be NARROWER than Question 1's - more data gives a more precise estimate
```

**4. SOLVED - Type I vs Type II error examples**

- Type I error example: a doctor tells a healthy patient they have a disease (false alarm - the test said 'positive' but the patient is actually fine).
- Type II error example: a doctor tells a sick patient they are healthy (missed finding - the test said 'negative' but the patient actually has the disease).
